# Different types of Arguments:
* Positional argument
* Keyword

## Positional Arguments:  

````python
def add(a,b):  
    print(a, b)
add(1,64) # here 1 and 64 are positional argument
````

````python
def add(*args):  
    print(args)
    for i in args:
        print(i)
    print(*args)
    
add(1,64,34,5,35)
````

## Keyword arguments:  


````python
def show(a,b,c):
    print(f"{a} {b} {c}")

show(a = 10, c = 30, b = 40) # These are key word arguments were the arguments is assigned
````

````python
def show(**args):
    print(args)
    for key, value in args.items():
        print(key, value)
    
show(a = 34, c = 53, b = 34) # it shows data in a dictionary attributes
````

# Generator: get data as a chunk size

````python
lst = [x for x in range(500)] # dataset

def data_loader(chunk_size, lst):
    for i in range(0,len(lst),chunk_size):
        yield lst[i: i+chunk_size:]

x = data_loader(5,lst)
print(type(x)) # the data type will give generator
print(x)
print(next(x))
print(next(x))
````

Write a Python generator function called batch_generator. It should take a list of numbers (representing data samples) and a batch_size. The generator should yield one batch at a time as a list.

Example Input:  

data = [1, 2, 3, 4, 5, 6, 7, 8]  

batch_size = 3


Expected Output (when iterated):  

[1, 2, 3]  

[4, 5, 6]  

[7, 8]


### solution code:
```python
def data_load(lst, batch_size):
    for i in range(0, len(lst), batch_size):
        yield lst[i:batch_size+i]

data = [int(x) for x in input("data = ").split()]
n = int(input("batch_size = "))

for i in data_load(data,n):
    print(i)
```

### Problem 4:

Write a function power(base, exponent) that returns the result of base raised to exponent. If the exponent is not given, it should calculate the square.  

Sample Input:  

5  

Sample Output:  

25  

Sample Input 2:  

5 3  

Sample Output 2:  

125


### solution code:
```python
def power(base, exponent = 2):
    result = 1
    for i in range(exponent):
        result *= base
    return result

nums = [int(x) for x in input().split()]
if len(nums) == 1:
    print(power(nums[0]))
else: print(power(nums[0],nums[1]))
```
