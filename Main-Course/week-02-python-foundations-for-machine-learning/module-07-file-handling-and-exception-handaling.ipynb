## File Handling in Python (This section is taken from ai(chatgpt) and these information helped me to learn these basic concepts)

### 1. Basic Syntax

```python
file = open("data.txt", "r")
content = file.read()
file.close()
```

**Better way:** Use `with` to close the file automatically.

```python
with open("data.txt", "r") as file:
    content = file.read()
```

### 2. File Modes

| Mode | Meaning |
|---|---|
| `"r"` | Read; file must exist |
| `"w"` | Write; creates or overwrites file |
| `"a"` | Append; adds content at the end |
| `"x"` | Create a new file; error if it exists |
| `"b"` | Binary mode, e.g. `"rb"` |
| `"t"` | Text mode (default) |
| `"r+"` | Read and write; file must exist |
| `"w+"` | Read and write; overwrites file |
| `"a+"` | Read and append |

### 3. Basic Operations

```python
file.read()        # Read entire file
file.readline()    # Read one line
file.readlines()   # Read lines into a list
file.write("Hi")   # Write a string
file.writelines(["A\n", "B\n"])  # Write multiple strings
```

### 4. File Handling Flowchart

```mermaid
flowchart TD
    A([Start]) --> B[Open file with mode]
    B --> C{Choose operation}
    C -->|r| D[Read]
    C -->|w| E[Write or overwrite]
    C -->|a| F[Append]
    D --> G[Close file]
    E --> G
    F --> G
    G --> H([End])
```

### 5. Important Notes

- `"w"` **deletes existing content** before writing.
- `"a"` adds content at the end.
- `with open(...)` closes the file automatically.
- `read()` returns a string; `readlines()` returns a list of strings.


````python
file = open("text.txt","r")
content = file.read() # read() function only read the file
print(f"{type(content)}\n{content}")

file.close() # close() function close the function and always we have to close the function we opened
print(file.closed) # return True if file if closed and return False if the file is opened
````

````python
with open("text.txt", "r") as file: # this is best habit declaring an file because it automatically close the file
    content = file.readlines() # readlines() function stores the file in a list
    print(content)
    content = list(map(str.strip, content)) # the "\n" is gone form the content list 
    print(content)
    content = list(filter(lambda x : len(x)>0,content)) # filter out the extra space
    print(content)


print(file.closed)
````

## File Pointer

````python
with open("sample.txt","r") as file:
    print(file.tell()) # tell() function tells where the cursor/ pointer is in the file currently
    content = file.read()
    print(content)
    print(file.tell())
    print(file.read())
````

````python
with open("sample.txt","r") as file:
    print(file.read(5)) # it read from starting cursor to 5 length
    print(file.tell())
    print(file.read())
````

````python
# Seek

with open("sample.txt","r") as file:
    print(file.tell())
    file.seek(5) # with seek(which index i want the cursor at) function we can move the cursor
    print(file.tell())
    print(file.read())
    file.seek(3)
    print(file.tell())
    print(file.read())
````

## Exception Handling in Python (This section is taken from AI (Chatgpt), and these information helped me to learn these basic concepts)

### 1. Basic Syntax

```python
try:
    num = int(input("Enter a number: "))
    print(10 / num)

except:
    print("Something went wrong!")
```

**Better way:** Handle specific exceptions instead of catching everything.

```python
try:
    num = int(input("Enter a number: "))
    print(10 / num)

except ValueError:
    print("Please enter a valid number.")

except ZeroDivisionError:
    print("Cannot divide by zero.")
```

### 2. Exception Handling Keywords

| Keyword | Meaning |
|---|---|
| `try` | Contains code that may cause an exception |
| `except` | Handles an exception |
| `else` | Runs if no exception occurs |
| `finally` | Always runs, whether an exception occurs or not |
| `raise` | Manually raises an exception |

### 3. Basic Structure

```python
try:
    num = int(input("Enter a number: "))
    result = 10 / num

except ValueError:
    print("Invalid input!")

except ZeroDivisionError:
    print("Cannot divide by zero!")

else:
    print("Result:", result)

finally:
    print("Program finished.")
```

- `try` → Attempts to execute the code.
- `except` → Handles matching exceptions.
- `else` → Runs only if no exception occurs in `try`.
- `finally` → Runs whether an exception occurs or not.

### 4. Common Exceptions

| Exception | Cause |
|---|---|
| `ValueError` | Invalid value conversion, e.g. `int("abc")` |
| `ZeroDivisionError` | Dividing by zero |
| `TypeError` | Incompatible data types |
| `IndexError` | Accessing an invalid list index |
| `KeyError` | Accessing a missing dictionary key |
| `FileNotFoundError` | Opening a nonexistent file |
| `NameError` | Using an undefined variable |

### 5. Exception Handling Flowchart

```mermaid
flowchart TD
    A([Start]) --> B[Execute try block]
    B --> C{Exception occurs?}
    C -->|No| D[Execute else block]
    C -->|Yes| E[Execute matching except]
    D --> F[Execute finally]
    E --> F
    F --> G([End])
```

### 6. Raising an Exception

Use `raise` to trigger an exception manually.

```python
age = -5

if age < 0:
    raise ValueError("Age cannot be negative.")
```

### 7. Important Notes

- An exception interrupts the normal flow of a program.
- Use specific `except` blocks to handle different errors.
- If no matching `except` exists, the exception propagates.
- `else` runs only when the `try` block completes without an exception.
- `finally` is useful for cleanup operations.
- Avoid bare `except:` unless you have a good reason.
- Exception handling helps prevent unexpected program termination when errors can be handled safely.
