## 🌟Methods of sets in python:  

```python

S = {1,2,6,3,6,6,4}
S.add(0)  #### Add function #####
S.add(35)
print(S)

S.pop() # pop function in set and it pop the first element in the set
print(S)

S.remove(35) # removing a specific element from the set
print(S) 
```

## 🌟 Set Mathematical methods:


```python
A = {1,5,6,6,3,7,8}
B = {3,5,6,7,8,1}

Union = A.union(B) # set union function. every element of A and B only one time
print(Union)

Intersection = A.intersection(B) # set intersection just common element 
print(Intersection)

Disjoint = A.isdisjoint(B) # This function returns True or False. Its check whether the set is disjoint or not . disjoint means there will be no common element
print(Disjoint) # in A and B set there is multiple common element so it returns False

Subset = A.issubset(B) # issubset return boolean . it's checking wheather A set is subset of B set meaning is making A set form B set is possible or not
print(Subset)
```

## 🌟 Dictionary Basics:

## Some propertise of Dictionary in Python:  
* A dictionary is a built-in Python data structure that stores data in (key, value) pairs.
* It is mutable
* It is an unordered data structure
* In  pair(key,value) the key have to be ummutable(ex: int, string, float, tuple etc) and value can be anything

```python
dic = {} # an empty dictionary
print(type(dic))
```

```python
info = {"name: ":"Dipu Mondol", "age: ":23, "address: ": "Natore"} # here name, age, address are keys and "Dipu", 23, Natore are values
print(info)

# accessing dictionay by keys 
print(info["name: "])
print(info["age: "])
print(info["address: "])

info["name: "] = "Dipto Mondol" # the name section i now overwritten by Dipto Mondol
print(info["name: "])
print(info)
```

## 🌟 Dictionary Methods 🌟

```python
print(info.get("math_marks")) # with get function if any key value is not present in the dictionay it will output "none"
print(info.get("math_marks", 0)) # we can get our intersted output with get function if the element is not present. like in this case it printed 0 instead of None

# adding element using update function
info.update({"math Marks: ": 57})
print(info)

# deleting elements form a dictionary
del info["math Marks: "]
print(info)

# copying an dictiory 
info_copy = info.copy()
print(info_copy)
```

## 🌟Iterating over Dictionary: 


```python
info = {"name: ":"Dipu Mondol", "age: ":23, "address: ": "Natore", "math_numbers: ":[135,34,56]}

keys = info.keys() # keys() function stores the keys in the dictionary
print(keys)

values = info.values() # values() function stores the values of the dictionary
print (values)

items = info.items() # we can access the keys and values in form of tuple with the items() function
print(items)

for key,value in info.items(): # for iterating over the dictionary items() function is necessary
    
    print(key,value)
```

## 🌟 Dictionary Comprehension:  

```python
# Dictionary Comprehension: {key: value for item in iterable if condition}
squre = {x:x**2 for x in range(1,6) if x**2 % 2 == 0}
print(squre)
```

```python

coordinates = [(10,10.5), (20.5, 192), (101, 102)]
locations = ["dhaka", "chattagram", "syllet"]

exact_location = {cor : loc for cor, loc in zip(coordinates, locations)}
print(exact_location)
```

## Problem solving with the built in data structure in python:

 ## 1. Given a list of numbers , make it a list with unique values

```python
lst = [10, 20, 10, 30, 30, 50, 30, 10, 20, 10, 10]

Set = set()

for x in lst:
    Set.add(x)
    

new_lst = []
for x in Set:
    new_lst.append(x)
print(new_lst)
```

```python
#### Alternative solution @@@@
lst = [10, 20, 10, 30, 30, 50, 30, 10, 20, 10, 10]
Set = set(lst)
new_lst = list(Set)

print(new_lst)
```

## 2.Given a strign print the frequency of the words

```python
string = """ Python is easy to learn. 
Python is powerful and
Python is popular. """

words = string.split()
freq_lst = [0 for _ in range(len(words))]

ans_dic = {keys:values for keys, values in zip(words,freq_lst)}


for i in words:
    ans_dic[i] += 1
for keys,values in ans_dic.items():
    print(keys,values)
    
```

```python
#### Alternative solution #####

temp_string = """ Hi, I'm Dipu Mondol.
I am 23 years of old.
I am currently studying python """

word_split = temp_string.split()

dic = {}

for x in word_split:
    dic[x] = dic.get(x,0) + 1


for x,y in dic.items():
    print(x,y)





```

## Practice Problem 1: [The Link](https://codeforces.com/group/MWSDmqGsZm/contest/219856/problem/J)

## The solution:

```python
string = input()
dic = {}
for x in string:
    dic[x] = dic.get(x,0) + 1

ans_dic = dict(sorted(dic.items())) # makes the dictionary forted
for keys, values in ans_dic.items():
    print(f"{keys} : {values}")
```
