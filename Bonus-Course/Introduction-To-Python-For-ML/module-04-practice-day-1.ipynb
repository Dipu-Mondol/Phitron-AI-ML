## Practice problem 2: [The Link](https://www.hackerrank.com/contests/ai-ml-module-4-practice-day-1/challenges/b3-sensor-calibrator-min-max-scaling/problem?isFullScreen=true)

## Practice problem 3: [The Link](https://www.hackerrank.com/contests/ai-ml-module-4-practice-day-1/challenges/b4-mini-loss-tracker-average-monitor/problem?isFullScreen=true)  

## The solution: 



```python
N = int(input())
target = float(input())

total_loss = 0.0

for i in range(N):
    total_loss += float(input())

average = float(total_loss / N)

if average <= target:
    print("PASS")
else: print("RETRY")    
```

## Practice problem 4: [The Link](https://www.hackerrank.com/contests/ai-ml-module-4-practice-day-1/challenges/b5-majority-vote-gate-simple-ensemble/problem?isFullScreen=true)
## The solution:


```python
N = int(input())

Yes = 0
No = 0
for i in range(N):
    string = input()
    if string == "YES":
        YES += 1
    else: No += 1
if Yes >= No:
    print("ACCEPT")
else: print("REJECT")    
```

## Practice problem 5: [The Link](https://www.hackerrank.com/contests/ai-ml-module-4-practice-day-1/challenges/ai-mood-detector/problem?isFullScreen=true)
## The solution:

```python
string = input()

if string.find("happy")>-1 or string.find("joy")>-1 or string.find("smile")>-1:
    print("Happy Mood")
elif string.find("sad")>-1 or string.find("cry")>-1 or string.find("angry")>-1:
    print("Sad Mood")
else: print("Neutral Mood")
    
```

## Practice problem 6: [The Link](https://www.hackerrank.com/contests/ai-ml-module-4-practice-day-1/challenges/p2-a2-ai-brightness-analyzer/problem?isFullScreen=true)
## The solution:


```python
lst = [float(x) for x in input().split()]
total = 0.0
for x in lst:
    total += x
avarage = total/float(len(lst))

if avarage < 85:
    print("Dark Image")
elif avarage >= 85 and avarage <= 170:
    print("Normal Image")
else: print("Bright Image")
```

## Practice Problem 7: [The Link](https://www.hackerrank.com/contests/ai-ml-module-4-practice-day-1/challenges/p2-a3-ai-bias-detector/problem?isFullScreen=true)  
## The solution:  


```python
ch_lst = [x for x in input().split()]
count_A = 0.0
count_B = 0.0

for i in ch_lst:
    if i == "A":
        count_A += 1
    else: count_B += 1
ans = [count_A, count_B]
final = (max(ans) / float(len(ch_lst))) * 100.0
if final > 70.0:
    print("Biased Model")
else: print("Fair Model")
```

## Practice Problem 8: [The Link](https://www.hackerrank.com/contests/ai-ml-module-4-practice-day-1/challenges/p2-a4-ai-keyword-counter/problem?isFullScreen=true)  
## The solution: 


```python
string = input()
count = 0
lst = ["ai", "data", "model", "learn", "train", "neural"]
temp = string.split() # This is use because in training there is train and ai both word so it count 2 although it is a one word
for x in lst:
    count += temp.count(x)
if count >=2 :
    print("AI Detected")
else : print("Not AI Related")
```
