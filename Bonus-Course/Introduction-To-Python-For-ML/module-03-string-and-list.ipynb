### 🧑🏻‍💻String Methods:


```python
String = "Hi, I'm Dipu Mondol. dipu is a moron. moron dipu moron boy"

processed_string = String.lower() # now with the lower() function all ther letter in the string became small latter
print(processed_string)

processed_string = String.capitalize() # with capitalize() function the fist latter of string now capital.
print(processed_string)

length = len(String) # len() function for findig lenght 
print(length)

print("Dipu" in String) # in function for checking if the function is in the String
print("Dipu" in processed_string)

idx = processed_string.find("dipu") # find() function for finding Starting index of dipu sub_string
print(idx)
reverse_idx = processed_string.rfind("dipu") # rfind() function for finding dipu substring firom the end
print(reverse_idx)

count = processed_string.count("dipu") # count() function for counting the number of subtring in the main string
print(count)

new_string = String.replace("moron", "inteligent") # replace() function replaced the moron substring with inteligent
print(new_string)
```

### 🌟List and List Methods

```python
numbers = [1,53445,545,6]
print(f"The first number: {numbers[0]}.\nThe second number: {numbers[1]}")

new_list = numbers[1:3] # copying the list in range [start, end-1]
print(new_list)

numbers.append(53)
print(numbers)

numbers.insert(2, 34) # insert function that insert in insert(position, element) . It is called insertion at any position
print(numbers)

numbers.pop()  # with pop function last element is deleted and return the deleted element
print(numbers)

numbers.remove(53445) # with remove function(variable.remove(element) the selected element can bd removed . This function only delete the first selsected element
print(numbers)

names = ["Dipu", "Dipto", "RAhim", "Karim"]
print(names)

student_marks = [["Dipu", "Diputo", "Rahim",45.9],[12,35,53], [14,543,67], [54,645,35]] # list within list with different datatypes
print(student_marks)
print(student_marks[1][2]) # printing by index
print(student_marks[0][3])
print(type(student_marks[0][3]))

numbers.sort() # the numbers list is sorted in accending order
print(numbers)
numbers.sort(reverse = True) # sorting number list in decending order
print(numbers)
```

### 🌟 List as Stack and Queue:


### Stack as list: LIFO(Last In First Out)

```python
stack = [] # empty list named stack
stack.append(1)
stack.append(3)
stack.append(5)
stack.append(6)
print(stack)

stack.pop() # poped the last element is poped first which inserted in the last
stack.pop()
print(stack)
print(stack[len(stack)-1], "or" , stack[-1]) # printing the top element
```

### Queue as list: FIFO(First in First Out)

```python
queue = []

queue.append(1)
queue.append(2)
queue.append(3)
queue.append(5)
print(queue)

##### Remove the first element #####
queue.pop(0)
queue.pop(0)
queue.pop(0) # When the first element is removed, the remaining elements automatically shift one position to the left.
print(queue)
```

## 🌟List Comprehension

```python
random_list = [x+10 for x in range(1,11) if x % 2 == 0]  # for each x in range of 1 to 10 if x is even than add 10 and put it in the list
print(random_list)
print(type(random_list[1]))

names_list = ["Rahul", "dipu", "Amit", "Dipu", "Rohit", "DIpU"] # A list of different name strings

matched_names = [name.upper() for name in names_list if name == "Dipu" or name == "dipu"] # List comprehension to find "Dipu" or "dipu" (case-sensitive)

print(matched_names)
```

### Practice problem 2: [The link](https://codeforces.com/group/MWSDmqGsZm/contest/219774/problem/M)
### The solution: 

```python
t = int(input())

array_list = [int(x) for x in input().split()]

maxi = -(10**6)
maxi_idx = 0
count_mx = 0

mini = 10**6
mini_idx = 0
count_mn = 0

for x in array_list:
    if x > maxi: 
        maxi = x
        maxi_idx = count_mx
    count_mx += 1    

for x in array_list:
    if x < mini:
        mini = x
        mini_idx = count_mn
    count_mn += 1    

array_list[maxi_idx] = mini
array_list[mini_idx] = maxi

print(*array_list)
```
