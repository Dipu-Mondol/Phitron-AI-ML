# Variables  

----

## Variable naming rules: 
* A variable name contains English letters (A-Z, a-z), digits (0-9), and underscores (_)
* A variable must start with an English letter or underscore, and a digit can be in the name after the latter or underscore
* Python is a case-sensitive language, meaning Dipu and dipu are two different variables
* Special keywords like if, else, for, while, etc. cannot be variable name
* Examples: 
  * Valid variable names: dipu_mondol, _dipuMondol, dipu40, dipu_40, NUM, Num
  * Invalid variable names: 12dipu, $dipu, dipu$mondol, else, dipu-mondol

```python
phone_num = 1786527071 # phone_num is a variable
print(phone_num)
```

# Data Types

----

## Some data types: 
* int (integer numbers like 20 , 30, 0, -10 etc)
* float (floating numbers like 1.2, 0.0, -3.5)
* str (strings like "Dipu Mondol", "I am hungry")
* bool (boolean like True or False)

```python
name = "Dipu Mondol"
age = 22
height = 5.6
is_passed = True

print(type(name), type(age), type(height), type(is_passed)) # here type is a function that tells which datatype is my variable
```
