# Question 1:

Given a list of integers, extract all numbers that are divisible by 4 but not divisible by 8.

Input Format:
Space-separated integers

Output Format:

A list of numbers satisfying the condition

Example:

Input:

4 6 8 12 16 20 24 28 


Output:

[4, 12, 20, 28] 



## Answer to the question No.: 01

```python
input_lst = [int(x) for x in input().split()]
result_lst = list(filter(lambda x : x%4 == 0 and x%8 != 0, input_lst))
print(result_lst)
```

# Question 2:
Write a function compute_ratio(x, y) that returns (x + y) / (x * y). Use exception handling to manage division by zero.
Use exception handling to manage:
Division by zero

Input Format:

Two values


Output Format:

Computed result OR error message
Example:

Input:

5 0

Output:

Error: Division not possible



## Answer to the question No.: 02  

```python
def compute_ratio(x,y):
  try:
   return (x + y) / (x * y)
  except ZeroDivisionError:
    return "Error: Division not possible"
  else:
    return (x + y) / (x * y)
a,b = input().split()
a = int(a)
b = int(b)
result = compute_ratio(a,b)
print(result)
```

# Question 3:
Write a Python program that reads a text file and finds the longest word.

Requirements:

Open and read data from a file (Example: data.txt)
If multiple words have the same longest length, print any one of them


Input Format:

A text file containing multiple lines of text


Output Format:

The longest word


Example File Content (data.txt):

Python is a wonderful language for machine learning.


Output:

wonderful



## Answer to the question No.: 03

```python
from functools import reduce
with open("data.txt","w+") as file:
  file.write("Python is a wonderful language for machine learning.")
  file.seek(0)
  string = file.read()
  temp_list = string.split()

  first_len = temp_list[0]
  for i in temp_list:
    if len(first_len) < len(i):
      first_len = i
  print(first_len)
```

# Question 4:
Write a program that converts a sentence into a list of words but only includes words that start with a vowel (a, e, i, o, u), ignoring case.

Input Format:

A single line sentence

OutputFormat:

A list of strings

Example:

Input:

An apple a day keeps the doctor away

Output:

['An', 'apple', 'a', 'away']



## Answer to the question No.: 04

```python
string = input()
str_lst = string.split()

str_lst = list(filter(lambda x : x[0] == 'a' or x[0] == 'e' or x[0] == 'i' or x[0] == 'o' or x[0] == 'u' or x[0] == 'A' or x[0] == 'E' or x[0] == 'I' or x[0] == 'O' or x[0] == 'U', str_lst))
print(str_lst)
```

# Question 5:
A word is called a Palindrome if reversing its letters results in the same word (ignoring case). Write a program to check whether a given string is a palindrome.

Input Format:
A single word string

Output Format:
"Palindrome" or "Not Palindrome"

Example:

Input:

Radar


Output:

Palindrome



## Answer to the question No.: 05

```python
def check_palindrome(string):
  temp = string[::-1]
  temp = temp.lower()
  string = string.lower()
  if temp == string:
    return True
  else: return False

string = input()
x = check_palindrome(string)
if x == True:
  print("Palindrome")
else: print("Not Palindrome")
```
