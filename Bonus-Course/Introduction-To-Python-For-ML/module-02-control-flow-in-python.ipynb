## Practice problem 2: [The Link](https://codeforces.com/group/MWSDmqGsZm/contest/219432/problem/C)

```python
#--------The solution of practice problem 2---------#

n = int(input())

nums = [int(x) for x in input().split()]   # n = int(input())
                                           # nums = [int(input()) for _ in range(n)]

even_count = 0; odd_count = 0; pos_count = 0; neg_count = 0

for i in range(n): 
    if nums[i] % 2 == 0:
        even_count += 1
    if nums[i] % 2 != 0:
        odd_count += 1
    if nums[i] > 0:
        pos_count += 1
    if nums[i] < 0: 
        neg_count += 1

print("Even: ",even_count)
print("Odd: ",odd_count)
print("Positive: ",pos_count)
print("Negative: ",neg_count)
```

```python
#--------Alternative solution of practice problem 2----------#

n = int(input()) 

inp = input()

numbers= inp.split()

positive =0 ; negative =0 ; even =0 ; odd =0 


for i in range(n):
    x = int(numbers[i])
    # positive or negative
    if x > 0 :
        positive+=1 
    elif x < 0 :
        negative+=1 
    
    # odd or even
    if x % 2 !=0 :
        odd+=1 
    else:
        even+=1 


print("Even:",even)
print("Odd:",odd)
print("Positive:",positive)
print("Negative:",negative)
```

## Practice problem 3: [The Link](https://codeforces.com/group/MWSDmqGsZm/contest/219432/problem/Q)

```python
#--------The solution of practice problem 3---------#

t = int(input())

while t != 0:
    n = input()
    nums = []
    length_of_n = len(n)
    n = int(n)
    for i in range(length_of_n):
        last_digit = n % 10
        nums.append(last_digit)
        n //= 10  # double division sign for getting integer part of the number unless it will give floating number
    
    print(*nums)
    t -= 1

    
```

```python
#-------Alternative solution of practice problem 3---------#

t = int(input())

while t > 0:
    n = int(input())
    if n == 0:
        print(n)
        t -= 1
        continue
    while n > 0: 
        print(n % 10, end = " ")
        n //= 10
    print()    
    t -= 1
```
